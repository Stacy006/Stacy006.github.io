```python
import pandas as pd
import altair as alt
```

```python
url = "https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/bfro_reports_fall2022.csv"
df = pd.read_csv(url)
df.head()
```

```python
df.info()
```

```python
df.columns
```

```python
df[['date', 'state', 'classification']].head(10)
```

```python
df['date'] = pd.to_datetime(df['date'], errors='coerce')

df['year'] = df['date'].dt.year

df[['date', 'year']].head(10)
```

```python
year_counts = df.dropna(subset=['year']).groupby('year').size().reset_index(name='count')

year_counts['year'] = year_counts['year'].astype(int)

year_counts.head(10)
```

```python
plot1 = alt.Chart(year_counts).mark_line(point=True).encode(
    x=alt.X('year:Q', title='Year'),
    y=alt.Y('count:Q', title='Number of Reports'),
    tooltip=['year:Q', 'count:Q']
).properties(
    title='Bigfoot Sightings by Year',
    width=650,
    height=350
)

plot1
```

```python
state_counts = df.groupby('state').size().reset_index(name='count')

state_counts = state_counts.sort_values('count', ascending=False)

state_counts.head(10)
```

```python
top10_states = state_counts.head(10)

plot2 = alt.Chart(top10_states).mark_bar().encode(
    x=alt.X('count:Q', title='Number of Reports'),
    y=alt.Y('state:N', sort='-x', title='State'),
    tooltip=['state:N', 'count:Q']
).properties(
    title='Top 10 States by Bigfoot Sightings',
    width=650,
    height=350
)

plot2
```

```python
selection = alt.selection_point(fields=['state'], on='click')

plot2 = alt.Chart(top10_states).mark_bar().encode(
    x=alt.X('count:Q', title='Number of Reports'),
    y=alt.Y('state:N', sort='-x', title='State'),
    color=alt.condition(selection, alt.value('orange'), alt.value('lightgray')),
    tooltip=['state:N', 'count:Q']
).add_params(
    selection
).properties(
    title='Top 10 States by Bigfoot Sightings',
    width=650,
    height=350
)

plot2
```

```python
plot1.save('plot1.json')
plot2.save('plot2.json')
```
